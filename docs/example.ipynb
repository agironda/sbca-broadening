```python
from sbca_broadening.broadening import geometric_broadening

results = geometric_broadening(
    source_x=0.010,
    source_y=0.010,
    source_z=0.010,
    diam_sbca=100,
    bending_radius=500,
    bragg=75,
    alpha=10,
    energy=7060,
    output="meV",
)

for group, values in results.items():
    if isinstance(values, dict):
        print(f"\n{group.upper()}")
        for name, value in values.items():
            print(f"  {name:10s}: {value:.1f} meV")
    else:
        print(f"\nOVERALL     : {values:.1f} meV")
```
